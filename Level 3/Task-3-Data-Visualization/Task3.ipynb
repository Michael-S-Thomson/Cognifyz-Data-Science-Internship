## Level 3 - Task 3: Data Visualization

### Objective

Create meaningful visualizations to understand restaurant ratings,
votes, price ranges, cuisines, and other important patterns in the dataset.

### Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Load the cleaned dataset

```python
df = pd.read_csv("Cleaned_Dataset.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nMissing values:")
display(df.isnull().sum())

print("\nDataset information:")
df.info()
```

### Visualize Restaurant Rating Distribution

## 1. Restaurant Rating Distribution

This visualization shows how restaurant ratings are distributed across the dataset.
It helps identify the most common rating ranges and understand the overall rating pattern.

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df["Aggregate rating"],
    bins=20,
    edgecolor="black"
)

plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")
plt.title("Distribution of Restaurant Ratings")

plt.tight_layout()
plt.show()
```

### Rating category counts

```python
rating_counts = (
    df["Aggregate rating"]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(12, 6))

plt.bar(
    rating_counts.index.astype(str),
    rating_counts.values
)

plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")
plt.title("Number of Restaurants by Aggregate Rating")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print("Rating distribution:")
display(rating_counts.to_frame(name="Restaurant Count"))
```

### Visualize Price Range Distribution

## 2. Price Range Distribution

This visualization shows the number of restaurants in each price range.
It helps understand the distribution of restaurants across different pricing levels.

```python
price_range_counts = (
    df["Price range"]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(8, 5))

plt.bar(
    price_range_counts.index.astype(str),
    price_range_counts.values
)

plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.title("Distribution of Restaurants by Price Range")

plt.tight_layout()
plt.show()
```

```python
print("Price range distribution:")
display(price_range_counts.to_frame(name="Restaurant Count"))
```

### Visualize Votes

## 3. Distribution of Customer Votes

This visualization shows how customer votes are distributed among restaurants.

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df["Votes"],
    bins=30,
    edgecolor="black"
)

plt.xlabel("Number of Votes")
plt.ylabel("Number of Restaurants")
plt.title("Distribution of Customer Votes")

plt.tight_layout()
plt.show()
```

### Visualize the Top 10 Cuisines

## 4. Top 10 Cuisines by Number of Restaurants

This visualization shows the most common cuisine types in the dataset based on the number of restaurants offering each cuisine.

```python
cuisine_df = df.assign(
    Cuisine=df["Cuisines"].str.split(", ")
).explode("Cuisine")

cuisine_counts = (
    cuisine_df["Cuisine"]
    .value_counts()
    .head(10)
)

print("Top 10 cuisines:")
display(cuisine_counts.to_frame(name="Restaurant Count"))
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    cuisine_counts.index[::-1],
    cuisine_counts.values[::-1]
)

plt.xlabel("Number of Restaurants")
plt.ylabel("Cuisine")
plt.title("Top 10 Cuisines by Number of Restaurants")

plt.tight_layout()
plt.show()
```


The visualization shows the most frequently represented cuisine types in the dataset.
North Indian cuisine has the largest number of restaurant entries, followed by Chinese,
Fast Food, and other cuisine categories.

### Visualize the Top 10 Cities

## 5. Top 10 Cities by Number of Restaurants

This visualization shows the cities with the highest number of restaurants in the dataset.

```python
city_counts = (
    df["City"]
    .value_counts()
    .head(10)
)

print("Top 10 cities:")
display(city_counts.to_frame(name="Restaurant Count"))
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    city_counts.index[::-1],
    city_counts.values[::-1]
)

plt.xlabel("Number of Restaurants")
plt.ylabel("City")
plt.title("Top 10 Cities by Number of Restaurants")

plt.tight_layout()
plt.show()
```

The visualization shows that restaurant distribution is concentrated in a small
number of cities. New Delhi has the largest number of restaurant entries in the
dataset, followed by other major cities.

### Visualize Online Delivery and Table Booking

## 6. Table Booking and Online Delivery

This visualization compares the availability of table booking and online delivery
services among restaurants.

```python
table_booking_counts = df["Has Table booking"].value_counts()

online_delivery_counts = df["Has Online delivery"].value_counts()

print("Table Booking:")
display(table_booking_counts.to_frame(name="Restaurant Count"))

print("\nOnline Delivery:")
display(online_delivery_counts.to_frame(name="Restaurant Count"))
```

### Table Booking visualization

```python
plt.figure(figsize=(7, 5))

plt.bar(
    table_booking_counts.index,
    table_booking_counts.values
)

plt.xlabel("Table Booking")
plt.ylabel("Number of Restaurants")
plt.title("Restaurant Availability of Table Booking")

plt.tight_layout()
plt.show()
```

### Online Delivery visualization

```python
plt.figure(figsize=(7, 5))

plt.bar(
    online_delivery_counts.index,
    online_delivery_counts.values
)

plt.xlabel("Online Delivery")
plt.ylabel("Number of Restaurants")
plt.title("Restaurant Availability of Online Delivery")

plt.tight_layout()
plt.show()
```

The visualizations show the availability of table booking and online delivery
services across restaurants in the dataset. These services are not equally
available across all restaurants, indicating differences in the services offered
by restaurants.

### Average Rating by Price Range

## 7. Average Rating by Price Range

This visualization compares the average restaurant rating across different price ranges.
It helps examine whether restaurants in different price categories have different average ratings.

```python
avg_rating_by_price = (
    df.groupby("Price range")["Aggregate rating"]
    .mean()
    .round(2)
)

print("Average rating by price range:")
display(avg_rating_by_price.to_frame(name="Average Rating"))
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    avg_rating_by_price.index.astype(str),
    avg_rating_by_price.values
)

plt.xlabel("Price Range")
plt.ylabel("Average Rating")
plt.title("Average Restaurant Rating by Price Range")

plt.ylim(0, 5)

plt.tight_layout()
plt.show()
```

The visualization compares average restaurant ratings across the four price ranges.
The results show differences in average ratings between price categories. Price Range 4
has the highest average rating in this dataset, with an average rating of approximately 3.82.

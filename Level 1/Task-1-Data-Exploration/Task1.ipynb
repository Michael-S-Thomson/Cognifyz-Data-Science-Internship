Import libraries:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Load the dataset:

```python
df = pd.read_csv("Dataset .csv")

df.head()
```

Find rows and columns:

```python
print("Dataset Shape:", df.shape)
```

Check the datatypes:

```python
df.info()
```

Display a summary of numerical columns:

```python
df.describe()
```

Check missing values:

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

Find the most common missing values:

```python
most_common_cuisine = df["Cuisines"].mode()[0]

print("Most common cuisine:", most_common_cuisine)
```

Fill the missing values:

```python
df["Cuisines"] = df["Cuisines"].fillna(most_common_cuisine)
```

Verify all missing values:

```python
remaining_missing = df.isnull().sum()

print("Remaining missing values:")
print(remaining_missing[remaining_missing > 0])
```

Missing Value Handling:

The dataset contained 9 missing values in the Cuisines column. Since Cuisines is a categorical variable, the missing values were replaced with the most frequently occurring cuisine. After treatment, no missing values remained in the dataset.

Check for Duplicate Records:

 Count duplicate rows:

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

Duplicate Check:

Duplicate records were checked in the dataset. No duplicate rows were found, so no duplicate records were removed.

Data Type Conversion:

 Check the current data types:

```python
print("Current data types:")
print(df[["Aggregate rating", "Votes", "Price range"]].dtypes)
```

Convert them to numeric:

```python
df["Aggregate rating"] = pd.to_numeric(
    df["Aggregate rating"],
    errors="coerce"
)

df["Votes"] = pd.to_numeric(
    df["Votes"],
    errors="coerce"
)

df["Price range"] = pd.to_numeric(
    df["Price range"],
    errors="coerce"
)
```

Verify the conversion

```python
print(df[["Aggregate rating", "Votes", "Price range"]].dtypes)
```

Data Type Conversion:

The numerical columns Aggregate rating, Votes, and Price range were checked and converted to appropriate numerical data types where necessary. The resulting data types were verified after conversion.


Analyze Aggregate Rating Distribution:

Get basic statistics:

```python
print(df["Aggregate rating"].describe())
```

Count each rating:

```python
rating_counts = df["Aggregate rating"].value_counts().sort_index()

print("Rating Distribution:")
print(rating_counts)
```

Calculate percentages:

```python
rating_percentage = (
    df["Aggregate rating"]
    .value_counts(normalize=True)
    .sort_index() * 100
)

print("Rating Distribution (%):")
print(rating_percentage.round(2))
```

Identify the most common rating:

```python
most_common_rating = df["Aggregate rating"].mode()[0]
most_common_count = df["Aggregate rating"].value_counts().max()

print("Most common rating:", most_common_rating)
print("Number of restaurants:", most_common_count)
```

Check the zero-rating group:


```python
zero_rating_count = (df["Aggregate rating"] == 0).sum()
zero_rating_percentage = (
    zero_rating_count / len(df)
) * 100

print("Restaurants with rating 0:", zero_rating_count)
print("Percentage with rating 0:", round(zero_rating_percentage, 2), "%")
```

Aggregate Rating Analysis:

The distribution of aggregate ratings was analyzed using descriptive statistics, frequency counts, percentages, and visualizations. The analysis shows that restaurants are not evenly distributed across rating values. A substantial proportion of restaurants have an aggregate rating of 0, indicating an imbalanced rating distribution. This imbalance should be considered when performing further statistical or predictive analysis.

Final validation :

```python
print("========== FINAL DATASET CHECK ==========")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("=========================================")
```
